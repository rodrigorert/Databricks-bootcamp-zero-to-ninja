```python
from pyspark.sql.functions import current_timestamp, date_format
from pyspark.sql.functions import col
dbutils.widgets.removeAll() #para que elimine todos lo valores almacenado y no tomer valores antiguos
```

```python

dbutils.widgets.text("source_table", "samples.bakehouse.sales_suppliers")
dbutils.widgets.text("target_table", "bronze_dev.bakehouse.suppliers_brz")

#"samples.bakehouse.sales_customers" #"bronze_dev.bakehouse.customer_brz"
#samples.bakehouse.sales_franchises #bronze_dev.bakehouse.franchises_brz
#samples.bakehouse.sales_suppliers #bronze_dev.bakehouse.suppliers_brz
#samples.bakehouse.sales_transactions # bronze_dev.bakehouse.transactions_brz
```

```python
source =dbutils.widgets.get("source_table") 
target = dbutils.widgets.get("target_table") 

df = spark.read.table(source)
```

```python
# Automatiza la selección y casteo de columnas para cualquier tabla


def select_and_cast_all_columns(df):
    cols = [col(c).cast('String').alias(c) for c in df.columns]
    cols += [
        current_timestamp().alias('ingest_timestamp'),
        current_timestamp().cast("date").alias('ingest_date')
    ]
    return df.select(*cols)

df_final = select_and_cast_all_columns(df)
```

```python
#guarder tabla
df_final.write.mode("overwrite").saveAsTable(target)
```
